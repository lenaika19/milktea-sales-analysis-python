```python
import numpy as np
import pandas as pd
```

```python
branches = pd.read_csv('branches.csv')
menu = pd.read_csv('menu.csv')
transactions = pd.read_csv('transactions.csv')
```

```python
branches
```

```python
branches_clean = branches.copy()
```

```python
branches_clean.duplicated().sum()
```

```python
branches_clean.info()
```

```python
branches_clean['region'] = (
    branches_clean['region']
    .str.strip()
    .replace({
        'ncr': 'NCR'
    })
)
```

```python
branches_clean
```

```python
branches_clean['opened_date'] = pd.to_datetime(
    branches_clean['opened_date'],
    format='mixed',
    errors='coerce'
)
```

```python
branches_clean
```

```python
branches_clean.to_csv('cleaned_data/branches_clean.csv', index=False)
```

```python
menu
```

```python
menu_clean = menu.copy()
```

```python
menu_clean
```

```python
menu_clean.duplicated().sum()
```

```python
menu_clean.dtypes
```

```python
menu_clean['category'].unique()
```

```python
menu_clean['category'] = (
    menu_clean['category']
    .str.strip()
    .replace({
        'milk tea': 'MilkTea',
        'MilkTea': 'MilkTea',
        'Milk Tea': 'MilkTea',
        'fruit tea': 'Fruit Tea',
        'coffee': 'Coffee',
        'Snack': 'Snacks',
        'Addons': 'Add-Ons',
        'Add-ons': 'Add-Ons',
        'Add-ons': 'Add-Ons',
        'add-on': 'Add-Ons'
    })
)
```

```python
menu_clean
```

```python
menu_clean['price'] = pd.to_numeric(
    menu_clean['price']
    .astype('string')
    .str.replace('₱','',regex=False)
    .str.replace(',','',regex=False)
    .str.strip(),
    errors='coerce'
)
```

```python
menu_clean['cost'] = menu_clean['cost'].fillna(0)
```

```python
menu_clean
```

```python
menu_clean.to_csv('cleaned_data/menu_clean.csv', index=False)
```

```python
transactions
```

```python
transactions_clean=transactions.copy()
```

```python
transactions_clean
```

```python
transactions_clean.duplicated().sum()
```

```python
transactions_clean = transactions_clean.drop_duplicates()
```

```python
transactions_clean.duplicated().sum()
```

```python
transactions_clean
```

```python
transactions_clean.dtypes
```

```python
transactions_clean.isna().sum()
```

```python
transactions_clean['txn_datetime'] = pd.to_datetime(
    transactions_clean['txn_datetime']
    ,format='mixed'
    ,errors='coerce'
)
```

```python
transactions_clean
```

```python
transactions_clean['payment_method'].unique()
```

```python
transactions_clean['payment_method'] = (
    transactions_clean['payment_method']
    .astype('string')
    .str.strip()
    .replace({
        'card': 'Card',
        'Credit Card': 'Card',
        'cash': 'Cash',
        'CASH': 'Cash',
        'G-Cash': 'GCash',
        'GCASH': 'GCash',
        'gcash': 'GCash',
        'PayMaya': 'Maya',
        'maya': 'Maya'
    })
)
```

```python
transactions_clean.dtypes
```

```python
transactions_clean['customer_type'].unique()
```

```python
transactions_clean['customer_type'] = transactions_clean['customer_type'].fillna('No Membership')
```

```python
transactions_clean['rating'] = transactions_clean['rating'].fillna('No Rating')
```

```python
transactions_clean['unit_price'] = pd.to_numeric(
    transactions_clean['unit_price']
    .astype('string')
    .replace('₱','',regex=False)
    .replace(',','',regex=False)
    .str.strip(),
    errors='coerce'
)
```

```python
menu_clean
```

```python
transactions_clean = transactions_clean[transactions_clean['qty'] > 0]
```

```python
transactions_clean = transactions_clean[
    transactions_clean['branch_id'].isin(branches_clean['branch_id']) &
    transactions_clean['item_id'].isin(menu_clean['item_id'])
]
```

```python
price_lookup = menu_clean.set_index('item_id')['price']
cost_lookup = menu_clean.set_index('item_id')['cost']
```

```python
transactions_clean['unit_price'] = transactions_clean['unit_price'].fillna(transactions_clean['item_id'].map(price_lookup))
```

```python
transactions_clean
```

```python
transactions_clean['unit_cost'] = transactions_clean['item_id'].map(cost_lookup)
```

```python
transactions_clean['line_revenue'] = transactions_clean['qty'] * transactions_clean['unit_price']
```

```python
transactions_clean['line_expense'] = transactions_clean['qty'] * transactions_clean['unit_cost']
```

```python
transactions_clean['line_profit'] = transactions_clean['line_revenue'] - transactions_clean['line_expense']
```

```python
transactions_clean
```

```python
transactions_clean.to_csv('cleaned_data/transactions_clean.csv',index=False)
```
