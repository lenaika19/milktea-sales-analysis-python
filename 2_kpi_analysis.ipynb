```python
import pandas as pd
import numpy as np
```

```python
transactions = pd.read_csv('cleaned_data/transactions_clean.csv')
branches = pd.read_csv('cleaned_data/branches_clean.csv')
menu = pd.read_csv('cleaned_data/menu_clean.csv')
```

```python
transactions['txn_datetime'] = pd.to_datetime(transactions['txn_datetime'])
```

```python
transactions
```

```python
menu
```

```python
fact = transactions.merge(
    menu[['item_id', 'item_name','category']],
    on='item_id',
    how='left'
)
```

```python
branches
```

```python
fact = fact.merge(
    branches[['branch_id', 'branch_name', 'region']],
    on='branch_id',
    how='left'
)
```

```python
fact
```

```python
fact.to_csv('fact_table.csv',index=False)
```

```python
total_revenue = fact['line_revenue'].sum()
```

```python
total_transactions = fact['txn_id'].nunique()
```

```python
average_per_transactions = total_revenue / total_transactions
```

```python
items_sold = fact['qty'].sum()
```

```python
total_profit = fact['line_profit'].sum()
```

```python
profit_margin = (total_profit/total_revenue)*100
```

```python
profit_margin
```

```python
print(f"Total Revenue: ₱{total_revenue:,.2f}")
print(f"Total Transactions: {total_transactions:,}")
print(f"AOV: ₱{average_per_transactions:,.2f}")
print(f"Items Sold: {items_sold:,}")
print(f"Profit Margin: {profit_margin:,.1f}%")
```
